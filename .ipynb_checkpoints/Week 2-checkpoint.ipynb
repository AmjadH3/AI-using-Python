## Topic 59 Handling Missing Data

 - data from client not ideal - missing values
 - 

```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv(
    "retail_sales.csv",
    parse_dates=["Date"],
    dtype={"Category":"category",
           "Region":"category"
          }
)

df.head()
```

```python
df.info()
```

```python
# Detect missing values
# .isna() shows null value (na value)
# .sum() shows sum of all null values in a column
print(df.isna().sum())
```

```python
# 2 ways to handle , 1- drop those rows (remove) 2- fill them (prefereable

# stastical use - for numerical cols use mean (average), median(midle value)
# for categorical cols use mode(frequent number), 
```

```python
# Fill missing age with mean
df['Quantity'] = pd.to_numeric(df['Quantity'])
df['Quantity'] = df['Quantity'].fillna(df['Quantity'].mode()[0])  # with mode it will take lot of vaues but we will use [0] index value for our null values
```

```python
# Fill missing value with median
df['Sales'] = df['Sales'].fillna(df['Sales'].median())
```

```python
# Fill categorical value with mode
df['Category'] = df['Category'].fillna(df['Category'].mode()[0])
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.dropna(subset = ['Region'], inplace=True) # inplace=true means to make changes in original file
```

```python
df.isna().sum()
```

## Topic 62 - Merging and Joining Dataframes
### Join -
    is similar to how tables are combined in databases
    - Inner join 
    - Left Join
    - Right Join
    - Outter Join

For GIS, this is extremely important. For example, you might have a shapefile containing District_ID and a CSV containing population data. You can use a left join to attach the population data to your district polygons while keeping every district geometry.

```python
import pandas as pd
```

```python
customers = pd.DataFrame({
    'CustomerID' : [1,2,3],
    'Name' : ['Alice', 'Bob', 'Charlie']
})

orders = pd.DataFrame({
    'OrderID' : [ 101, 102, 103, 104],
    'CustomerID' : [1,2,2,3],
    'Amount' : [250, 150, 200, 300]
})
```

```python
# Inner merge on CustomerID
df_inner = pd.merge(customers, orders, on = 'CustomerID' , how = 'inner')
print ("Inner Merge:\n", df_inner)
```

```python
#left merge
df_left = pd.merge (customers, orders , on = 'CustomerID', how = 'left')
print("Left merge:\n", df_left)
```

```python
#Concatenate rows example - when non related data in diff dataframes use concatenate
df_concat = pd.concat([customers, customers])
print("\nConcatenate Rows: \n", df_concat)
```

## Topic 63 - Applying Functions - apply, map & pipe
 - Through map and apply manipulate columns to make new ones
 - 

```python
df = pd.read_csv ("ecommerce_sales_analytics_5000.csv", parse_dates = ["order_date"])
```

```python
# varify whether we have correct data file
df.info()
```

```python
# map() - label high value orders, a col can be treated as series
# map() = go through each value and transform it.
df ["payment_type"] = df["payment_method"].map({
    "Card" : "Digital",
    "Wallet": "Digital",
    "COD" : "Cash"
})

# this code show that we introduce a new attribute using payment_method, 
# If payment was done by card its is digital, cod is cash etc
# Each values is mapped individually -> fast and simple
```

```python
# varify and check unique values 
df["payment_type"].unique()
```

```python
# print initial value using head function
df[['payment_method', 'payment_type']].head()     # Compare before and after
```

```python
# now how many dgital and how many cash payment types using size ().
df.groupby("payment_type").size()   # Business sanity check
```

```python
# Now Apply() function - create revenue category
# apply function can be applied to multiple cols, for complex criteria
# map () is little faster because it works with single series
def revenue_category(row):
    if row['revenue']>500:
        return 'Hight'
    elif row['revenue'] >= 200:
        return 'Medium'
    else:
        return 'Low'

df ['revenue_category'] = df.apply(revenue_category, axis = 1)  
# axis tells Pandas whether your function should work across rows or across columns.
# axis=0  → work DOWN the columns axis=1  → work ACROSS the rows( to each row)
# Why apply? Decision depends on multiple cols
```

```python
# varify
df['revenue_category'].value_counts()  
```

```python
df [['revenue', 'revenue_category']].sample(10)    # Validate logic manually
```

```python
# Now Pipe - stream line a series of operation like pipe line, sequence of operations
# pipe() - Method Chaining
# show that this sequence of methods should apply on that specific dataframes

def add_profit(data):
    data['profit'] = data['revenue']*0.20    # profit is 20% of revenue
    return data

def add_net_revenue(data):
    data['net_revenue'] = data['revenue'] - data['profit']   #we subtract profit from revenue to get ned revenue
    return data

df = (
    df
    .pipe(add_profit)
    .pipe(add_net_revenue)    # the above fuctions should be applied in same sequence as i wrote
)
```

```python
# Varifgy
df [['revenue', 'profit', 'net_revenue']].head()     # check new cols
```

## Topic 67 - Outlier Detection using Statistical Rules
- when load data some data doesnt match, not normal data range, fall outside normal data, extrame values can distort mean, medain.
### Idenditification and Removal
- IQR method
-  Z-Score method
-  Expert, domain- driven thresholds  
 99 % chances of outliers  
   qantiles probabalities like 25% probalilities 50 and 75%


- for time series data use IQR method
- for other datta use Z-SCORE METHOD

```python
import pandas as pd

df = pd.read_csv('ecommerce_sales_analytics_5000.csv', parse_dates = ['order_date'])
```

```python
# IQR Method
q1 = df['revenue']. quantile(0.25)
q3 = df['revenue']. quantile(0.75)

IQR = q3 - q1

lower_bound = q1 - 1.5 * IQR
upper_bound = q3 + 1.5 * IQR

outliers_iqr = df[(df['revenue'] < lower_bound) | (df['revenue'] > upper_bound)]

# IQR (Inerquantile Range): measure the middle 50% of data. Robust for skewed distributions (like revenue)
```

```python
outliers_iqr.shape    # Verification
```

```python
outliers_iqr[['revenue', 'product_category']].head()
```

```python
df = df[(df['revenue'] >= lower_bound) & (df['revenue'] <= upper_bound)]
df.shape
```

```python
#Z-Score Method
# HOw much our data deviate from stadard deviation, we keep data which are closer upto an extent and remove others
from scipy.stats import zscore

df = pd.read_csv('ecommerce_sales_analytics_5000.csv', parse_dates = ['order_date'])

df['revenue_z'] = zscore(df['revenue'])
outliers_z = df[df['revenue_z'].abs()> 3]

# Z-score shows: How many stadard deviations value is from the mean
# Rule : |z| > 3   outliers
# we can keep data upto 3 standard devaiton but can't keep bigger than that
```

```python
outliers_z.shape 
```

```python
df = df[df['revenue_z'].abs() <= 3]
df.shape
```

- If a varibale is increasing with another,
#### negative correlation
- If one var is increasing and 2nd is decreaing but both are related   


Not Causation - they don't affect each other

```python
# load Dataset
import pandas as pd
df = pd.read_csv('ecommerce_sales_analytics_5000.csv', parse_dates = ['order_date'])
```

```python
numeric_df = df[['quantity', 'unit_price', 'discount', 'delivery_days', 'revenue']]
# WE got all of numeric cols because covar and corre are of numeric cols only, not work on catergory and string cols
```

```python
# Covariance matrix- work on absolute values
# it will tell us the relations of variable with other in absolute units
print('\nCovariance:\n', numeric_df.cov())
```

```python
# Correlation - work on linear relation only bt -1 and +1, can also find ratio of how much related
print('\nCorrelation\n', numeric_df.corr())
```

```python
# Focus on revenue relationship
numeric_df.corr()['revenue'].sort_values(ascending=False)
```

```python
# Expected: min >= -1 and max <= =1

numeric_df.corr().values.min(), numeric_df.corr().values.max()
```

## Topic 69 - Creating an Exploratory Data Anaysis Report
* Base on all the functions applied (which we learned) we can create a report which is EDA report
* Heplful for client with data insights
#### Purpose of EDA report
* Summary key insights from the data
* Highlight data quality issues - if client should improve the data or not
* Identigy patterns, distributions and relationships
* Include tables, figures, and summary statistics - help client and our managment insight, varify if we understand data correctly( client is subject matter expert)
* provide checklist for further process

#### Activity Goal
* Draft a 6-8 bullet point summary of your hypothetical EDA findings for example a House Prices dataset
#### EDA Summary points
* Comment on data quality good quality good result e.g in age col 15% of records are missing (pin point problem is easy to solve), we can fill it through median if client wants
* Feature insights - data is left or right skewed, we will transform the data to minimize the skewedness or remove comlpletly
* Relation bt diff attributes - e.g sq feet of house and sale price,
* Categorical insight - eg top 3 category of property type col account for 80% of data like residential, commercial and parks
* Outliers - anomalies
* Temporal patterns - Time series eg sales tend to increase in last quarter of the year
### Key Takeaways
* EDA Report - narrative + visual evidence
* Use bullet points for clarity
* Include tables, figuresa nd key metrics
* Docs assumptions, anomalies and next steps
* Report should be concise and well structured
* Acts as a bridge bt raw and modeling data

#### Hands on Example 6-8 bullet eda summary
* Data Quality :  
  Arroud 8-10 % of record contains missing vlues in the customer rating col, good strategy to fil them if not then drop them
* Feature Insights:
  Revenue col is highly right-skewed, indicating that small number of large orders contribute disproportionaltly to total sales.  
  A Log transformation may be beneficial before applying machine learning models.
* Relationship:  
  Thire is a strong positive correlation (r ~ 0.82, this is 82% positive) bt quantitiy and revenue, confirming that higher order quantities directly drive highter revenue.
* Categorical Insight:  
  Top three product categories ( electronics, clothing, home) account for over 85% of total orders, indicating a strong concentration of sales in a limited number of categories.
* Outlier/Anomaly:  
   Sereral extreme outliers were observed in the revenue col, likely representing bulk or corporate orders. These outliers should be investigated rather than removed blindly, as they may carry important business value
* Temporal Patterns:  
   Sales show a clear upward trend in the final quarter of the year, suggesting seasonal effects likely driven by promotions, sales events, or increased consumer demand.


## Topic 70 - Matplotlib Figure Anatomy and Visualization Workflow
* Now if data is plot to visualize to better understand, we can show this to client to validate using matplotlib library.
* Matplotlib provides a figure that acts like canvas on which varous graphs and plots can be created.
* Graph are is called Axes:  
   x- axis and y- axis 2d  
   x axis  1d  
   x,y,z for 3d

#### Workflows
* Stateful (pyplot style) Functions operate on the current figure and axes (eg. plt.plot())  
   if single plot all operations will be applied on same plot
* Object oriented ( oo style):  
   Graph is oo style, uses explicit figure and axes aobjects for greater control, it will be a form of object on which diff methods will be called
#### Tips
* OO style is reusable and complex plots for clarity in code and in graph
* Lable the x and y axis clearly so viewers can understand easily
* Use descriptive label for plots
