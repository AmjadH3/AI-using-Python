```python
a = 4
row= a
col= a
matrix = []

for i in range(row):                         # Outer loop: runs 4 times (i = 0, 1, 2, 3)
    row = []                                 # Creates a NEW empty list for each iteration
    for c in range(col):                     # Inner loop: runs 4 times (c = 0, 1, 2, 3)
        row.append(0)                        # Adds 0 to the current row
    matrix.append(row)                       # Adds the completed row to the matrix

for row in matrix:                           # map(function, iterable) applies a function to every item in the iterable.
    print (' '.join(map(str, row)))
    #     └─┬─┘ └────┬────┘  └┬┘
    #       │        │        └─ the list of numbers
    #       │        └─ convert each to string
    #       └─ glue with spaces between
```

```python
customers = pd.DataFrame({
    'CustomerID' : [1,2,3],
    'Name' : ['Ishaq', 'Imran', 'Usman']
})

orders = pd.DataFrame ({
    'OrderID' : [12,13, 14,15],
    'CustomerID' : [1,2,2,3],
    'Amount' : [250,150,200,300]
})
```

```python
df_inner = pd.merge (customers , orders, on = 'CustomerID', how = 'inner')
print ('inner join\n :', df_inner)
```

```python
df_left = pd.merge(customers, orders, on= 'CustomerID', how = 'left')
print(df_left)
```

```python
pdconcat = pd.concat ([ customers, orders])
print(pdconcat)
```

```python
df = pd.read_csv("ecommerce_sales_analytics_5000.csv", parse_dates = ["order_date"])
df.info()
```

```python
df["payment_type"] = df["payment_method"].map({
    "Card" : "Digital",
    "Wallet" : "Digital",
    "COD" : "Cash"
})

```

```python
def revenue_category(row):
    if row['revenue'] > 500:
        return 'High'
    elif row['revenue'] >= 200:
        return 'Medium'
    else:
        return 'Low'

df['revenue_category'] = df.apply(revenue_category, axis = 1)
df['revenue_category'].value_counts()
```

```python
df[['revenue', 'revenue_category']].sample(10)
```

```python
def add_profit(data):
    data['profit'] = data['revenue'] * 0.20
    return data

def add_net_revenue(data):
    data['net_revenue'] = data['revenue'] - data['profit']
    return data

df = (
    df
    .pipe(add_profit)
    .pipe(add_net_revenue)

)
```
